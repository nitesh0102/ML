```python
Q1. What is Ridge Regression, and how does it differ from ordinary least squares regression?
```

```python
Ans: Ridge Regression is a type of linear regression that includes a regularization term to prevent overfitting, 
    particularly in situations where there is multicollinearity or when the number of predictor variables is large relative to the number of observations. It is a form of regularized regression,
    and it modifies the ordinary least squares (OLS) regression by adding a penalty to the size of the coefficients.
```

```python
Q2. What are the assumptions of Ridge Regression?
```

```python
Ans: 
1. Like OLS, ridge regression assumes that the relationship between the predictor variables (𝑋) and the target variable (𝑦) is linear.
2. The error terms (𝜖) should be independent of each other. This assumption implies that there is no autocorrelation in the residuals. In time series data, for example, the residuals should not show any systematic patterns over time.
3. The variance of the error terms (ϵ) should be constant across all levels of the independent variables. In other words, the residuals should have equal variance for all predictions (no heteroscedasticity).
4. Ridge Regression explicitly addresses the issue of multicollinearity 
5. While ridge regression can handle multicollinearity, it assumes that there is no exact collinearity in the predictors. In other words, no predictor should be a perfect linear combination of other predictors.
6.Normality of errors
7. No outliers in the predictors 
8. Appropriate choice of the regularization parameter (λ).
```

```python
Q4. Can Ridge Regression be used for feature selection? If yes, how?
```

```python
Ans : Ridge regression can be used for feature selection, but it does so differently from methods like Lasso regression. 
    Ridge regression does not set any coefficients exactly to zero, but it shrinks the coefficients of less important features toward zero. This means that while Ridge regression helps reduce the influence of less important features, it does not eliminate them entirely. In contrast, Lasso regression (another form of regularized regression) 
    can shrink some coefficients to exactly zero, which naturally leads to feature selection.
```

```python
Ans: Ridge Regression performs significantly better than Ordinary Least Squares (OLS) regression in the presence of multicollinearity.
```

```python
Ridge regression introduces a regularization term that shrinks the coefficients of correlated predictors.
```

```python
Q7. How do you interpret the coefficients of Ridge Regression?
```

```python
Interpreting the coefficients of Ridge Regression is similar to interpreting the coefficients of Ordinary Least Squares (OLS) regression, with the key difference being that Ridge Regression introduces a penalty (regularization) term that shrinks the coefficients. 
```

```python
Q8. Can Ridge Regression be used for time-series data analysis? If yes, how?
```

```python
Ans : Yes, Ridge Regression can be used for time-series data analysis, but with a few important considerations and adjustments to how you structure the data. While Ridge regression itself is a linear model with regularization, it does not inherently account for temporal dependencies or autocorrelations, which are key features of time-series data. However, 
    it can still be useful if you prepare the data correctly and handle the time-series structure appropriately.
```
